Simple Linear Regression

Importing the Libraries

```python

import pandas as pd
import matplotlib.pyplot as plt
```

Importing the dataset

```python
#importing dataset

dataset=pd.read_csv('Salary.csv') #reading the csv files and using it to dataset variable
x=dataset.iloc[:,:-1].values # indepenndent variable
y=dataset.iloc[:,1].values #dependent variable
```

Dataset

Splitting the dataset into the Training set and Test set

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size= .2, random_state = 0)
```

Training the Simple Linear Regression model on the trainig set

```python
from sklearn.linear_model import LinearRegression
regressor = LinearRegression()
regressor.fit(x_train, y_train) #relationship between x_train and y_train
```

Predicting the  Test set results

```python
y_predict = regressor.predict(x_test)
```

Visualising test Test set results

```python
plt.scatter(x_train, y_train, color = 'red')
plt.plot(x_train, regressor.predict(x_train), color = 'red')
plt.title('Salary vs Experience|Training set')
plt.xlabel('Years of Experience')
plt.ylabel('Salary')
plt.show()
```

Visualising the Test set results

```python
plt.scatter(x_test, y_test, color = 'blue')
plt.plot(x_train, regressor.predict(x_train), color = 'red')
plt.title('Salary vs Experience|Test set')
plt.xlabel('Years of Experience')
plt.ylabel('Salary')
plt.show()
```

Let's predict a salary of an employee with number of years of experience.

```python
years = input ("Enter the years of experience")
years_float = float (years)
prediction = regressor.predict([[years_float]])
print( "The salary of employee is:",prediction)
```
